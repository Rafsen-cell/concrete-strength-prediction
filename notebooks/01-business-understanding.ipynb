# Business Problem 
Concrete is one of the most widely used construction materials, and its compressive strength is a critical indicator of structural quality and safety. Traditionally, determining compressive strength requires laboratory testing after curing, which can take up to 28 days. This delay increase project time and cost when multiple concrete mix design need to be evaluated.

A predictive model can estimate the compressive strength of a concrete mix based on its ingredients and curing age, helping engineers evaluate potential mix designs before laboratory testing

# Business Objectives

- Develop a machine learning model to predict concrete compressive strength from mix design variables.
- Help engineers evaluate whether a concrete mix is likely to satisfy the required design strength.
- Reduce the number of unnecessary laboratory trials by identifying promising mix designs.
- Identify the material factors that have the greatest influence on concrete compressive strength.
- Build a simple decision-support application for estimating concrete strength.

# Analytical Questions
1. Which material properties have the strongest relationship with concrete compressive strength?
2. How do the input variables influence the predicted compressive strength?
3. Which regression model provides the most accurate predictions?
4. Which features contribute most to the model's predictions?
5. Can the predicted strength be translated into practical strength categories (e.g., Low, Medium, High) to support engineering decisions?

# Success Criteria
This project will be considered succesfull if it achieves the following:
- Build a reliable classification model with good predictive performance.
- Evaluate the model using confusion matrix, classification report, accuracy, precision, recall, f1-score, feature importance, model comparison table.
- Provide meaningful business insight from the analysis.
- Develop an interactive web application that allows engineers to imput concrete mix proportions and receive a predicted strength class
